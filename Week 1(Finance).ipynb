```python
# Machine Learning in Finance (Week1)
# My SID : 2533551   'Last 2 digits:51', 51 elements

import numpy as np

# Making a Maxtrix with 21 Elements
Matrix = np.arange(51)
print("Original Matrix :")
print(Matrix)

# 2D Matrix with 1 Row
Matrix= Matrix.reshape(1, -1)
print("\n 2D array with 1 row : ")
print(Matrix)

# Saving the original array to a different array
diffMatrix = Matrix
print("\n Coppied Matrix :")
print(diffMatrix)

# Shape of the diffMatrix ?
print("\n Shape of the diffMatrix")
print(diffMatrix.shape)
```
