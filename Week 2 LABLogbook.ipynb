```python
import pandas as pd
df = pd.read_csv("adult_data_mini-1.csv")
print(df.head()) 

# Grouping by "relationship" and "hours-per-week"
grouped_data = df.groupby(
    ["relationship", "hours-per-week"]
).size()
print(grouped_data)

# SID : 2533551
# Last digit of my SID : 1

n = 1
# Define a function to reduce hours
def reduce_hours(hours):
    return hours - n 

df["reduced-hours-per-week"] = df["hours-per-week"].apply(reduce_hours)       # Apply the function to create a new column
print(df[["relationship", "hours-per-week",
          "reduced-hours-per-week"]])
grouped_reduced_data = df.groupby(
    ["relationship", "reduced-hours-per-week"]         #Grouping by relationship and reduced hours
).size()

print(grouped_reduced_data)
```
